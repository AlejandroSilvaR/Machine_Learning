```python
import sys
import os
sys.path.append("../src")

import pandas as pd
import numpy as np
import joblib
import mlflow
import mlflow.sklearn

from config import *
from data import load_training_data, load_blind_data
from train import build_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X, y = load_training_data(TRAIN_DATA_PATH)

print("Dataset shape:", X.shape)
display(X.head())
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

model = build_model()
model.fit(X_train, y_train)

preds = model.predict(X_val)

rmse = mean_squared_error(y_val, preds)
r2 = r2_score(y_val, preds)

print("Validation RMSE:", rmse)
print("Validation R2:", r2)
```

```python
mlflow.set_experiment(MLFLOW_EXPERIMENT)

with mlflow.start_run():
    mlflow.log_param("model_type", MODEL_CONFIG["model_type"])
    mlflow.log_metric("rmse", rmse)
    mlflow.log_metric("r2", r2)
    mlflow.sklearn.log_model(model, "model")

print("MLflow run completed.")
```

```python
os.makedirs("../model", exist_ok=True)
joblib.dump(model, MODEL_PATH)

print("Model saved at:", MODEL_PATH)
```

```python
blind_df = load_blind_data(TEST_DATA_PATH)

blind_preds = model.predict(blind_df)

result_df = pd.DataFrame({"target_pred": blind_preds})
result_df.to_csv(RESULTS_DATA_PATH_DEMO_NOTEBOOK, index=False)

print("Predictions saved to:", RESULTS_DATA_PATH_DEMO_NOTEBOOK)
display(result_df.head())
```
